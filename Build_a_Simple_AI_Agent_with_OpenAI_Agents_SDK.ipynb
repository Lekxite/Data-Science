### Let's build and run our first AI agent using OPENAI AGENT SDK
### We will use Agent class from the openai-agents library:
### Creating an agent is like hiring an employee. You need to provide a job description. The key parameters are:

### Client: The OpenAI client we just configured. This is the agent's connection to the AI's brain such as ChatGPT, Claude, Gemini etc
### Name: A simple name to identify your agent (Fact Checker)
### Instructions: This is the most important part. It is the "System Prompt" or the detailed job description for the agent. This tells the agent who to be and what to do.
### Model: The specific OpenAI model the agent will use. (gpt-4.0-mini, is fast and cost effective or use the latest gpt-5 for enhanced results and longer context).

## **Prompt Components**
### Context: Background knoledge to help the model understand the context eg,"Assume that you are an expert financial analyst"
### Instruction: A clear statement of the desired goal. eg, "classify the following text as positive, neutral or negative
### Input Data: The data the model needs to process to produce. "Text" Apple stock increased by 2% in today's trading session.
### Output Indicator: Tells the model the format of the output. "Sentiment"

```python
# A function used to show the given text using markdown formatting in a Jupyter Notebook
def print_markdown(text):
    """Displays Markdown as text in Jupyter Notebook."""
    display(Markdown(text))
```

```python
# Import the OpenAI API Client
from openai import OpenAI
```

```python
# Import the Agent class to create and manage AI agents
from agents import Agent
```

```python
# Define the instructions for the fact-checker AI agents
fact_checker_instructions = """
Context:
You are a fact-checker who verifies the accuracy of statements.

Instructions:
When given a statement, carefully analyze its factual accuracy using your knowledge.

Input:
You will receive a statement that requires fact-checking.

Output:
Respond with:
1. A verdict prefix: either "TRUE:" or "FALSE:"
2. A brief, one-sentence explanation justifying your conclusion
"""
```

```python
# Create a new agent called "Fact-Checker"
fact_checker_agent = Agent(name = "Fact Checker", # The name of the Agent
                          instructions = fact_checker_instructions, # The rules and behaviours for the agent created above
                          model = "gpt-4.1-mini") # The AI model (LLM) to use.

# Print a configuration message that the agent was created
print(f"Agent '{fact_checker_agent.name}' created successfully!")
```

### Let's put our "Fact_Checker" agent to work! We will use the Runner.run() method to send it a statement to fact-check. The agent will analyze the statement and return its verdicts along with a brief explanation.

```python
# Import a Runner class, which is used to run an agent and get its output.
from agents import Runner
```

```python
# A statement we want the Fact Checker Agent to verify.
statement = "The great wall of China is visible from space with the naked eye"
```

```python
# Display the statement we are going to check (in markdown format for nicer formatting)
print_markdown(f"Asking the Fact Checker to verify: '(statement)'")
```

```python
# Run the Fact Checker agent on the input statement
# Await is used because it might take time to run and execute
response = await Runner.run(
    starting_agent = fact_checker_agent, # The agent we created earlier
    input = statement # The statement we want it to fact-check
)

# Display the agent's response
print_markdown("\n Agent's Response: \n")
print_markdown(response.final_output) # It shows the final verdict and explanation.
```

### **Build an AI Agent with Memory**

```python
# Import Required Classes from the Agents Module
# Agent: Represents an AI agent with a specific role and instructions
# Runner: Executes the agent and handles interactions
# SQLiteSession: Provides persistent memory storage for the agent using SQLite

from agents import Agent, Runner, SQLiteSession

# Define the Role and Instructions for the AI Agent
market_researcher_instructions = """
Context:
You are a market research assistant helping analyze companies, industries, and competitors.

Instructions:
When given a question, provide a short factual answer based on your knowledge.

Output:
Start with a prefix "FACT:" or "UNKNOWN:"
followed with a consice one-sentence explanation
"""

# Create an instance of the Agent
market_researcher_agent = Agent(name = "Market Researcher",
                                instructions = market_researcher_instructions,
                                model = "gpt-4.1-mini")
# Note: This one hasn't been configured with any memory yet
```

```python
# Let's give our first question to the AI agent
question1 = "What is the market share of Tesla in the US EV market?"

# Display the user's question in markdown format
print_markdown(f"You: '{question1}'")

# Run the agent with the first question (no memory means each query is independent)
response1 = await Runner.run(starting_agent = market_researcher_agent, input = question1)

# Display the agent's response
print_markdown(f"Agent: '{response1.final_output}'")
```

```python
# Second question-Depends on previous context
# Follow-up response that refers to previous answer
question2 = "How does that compare to last year?"

# Display the follow-up question
print_markdown(f"\nYou: '{question2}'")

# Run the agent again-since there is no memory, it does not recall the first question/answer.
response2 = await Runner.run(starting_agent = market_researcher_agent, input = question2)

# Display the agent's response (will fail to connect it to the first question)
print_markdown(f"Agent: '{response2.final_output}'")
```

### **Adding Memory to Agent using SQLiteSession**

```python
# Create a session instance
# SQLite-based implementation of session storage
# This implementation stores conversation history in a SQLite database

session = SQLiteSession("conversation")

# First Interaction
user_input1 = "What is the current share of Tesla in the US EV market?"
response1 = await Runner.run(
    starting_agent = market_researcher_agent,
    input = user_input1,
    session = session,
    )

print_markdown(f"Agent:\n{response1.final_output}")
```

```python
# Second Interaction with History
user_input2 = "How does that compare to last year?"
response2 = await Runner.run(
    starting_agent = market_researcher_agent,
    input = user_input2,
    session = session,
    )

print_markdown(f"Agent:\n{response2.final_output}")
```

## **Practice Question**
### Using OpenAI Agents SDK, create a new AI agent named Travel Planner that always suggests one sunny (warm) weekend getaway destination.
### 1. Write the agent's instructions; tell it to only give one sunny/warm destination.
### 2. Use the latest gpt-5-mini model.
### 3. Save the conversation history with SQLiteSession so it remembers past user questions and doesn't repeat destinations.
### Ask the agent: "Please suggest one weekend destination within 5 hours flying from Toronto, Canada.
### 5. Display the agent's answer.
### 6. Then ask a follow-up question about Visa requirements for Canadians for that destination and display the answer

```python
from agents import Agent, Runner, SQLiteSession

# Define the Role and Instructions for the AI Agent
travel_planner_instructions = """You an assistant specializing in weekend travel planning. Always suggest exactly ONE sunny (warm) destination
for a quick getaway
"""
```

```python
# Create an instance of the Agent
travel_planner_agent = Agent(name = "Travel Planner",
                                instructions = travel_planner_instructions,
                                model = "gpt-5-mini")
```

```python
session = SQLiteSession("travel_planner_session")

travel_msg = """Please suggest one weekend destination within 5 hours flying from Toronto, Canada."""

response1_travel = await Runner.run(
    starting_agent = travel_planner_agent,
    input = travel_msg,
    session = session,
    )

print_markdown(f"Travel Planner:\n{response1_travel.final_output}")
```

```python
followup_travel_question = "What are the Visa requirements for Canadians to this destination?"

response2_travel = await Runner.run(
    starting_agent = travel_planner_agent,
    input = followup_travel_question,
    session = session,
    )

print_markdown(f"Travel Planner:\n{response2_travel.final_output}")
```
